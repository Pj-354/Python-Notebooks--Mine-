```python
import numpy as np
import pandas as pd
```

```python
df_returns = pd.read_csv('Datasets\Top 11 NASDAQ Stocks.csv', parse_dates = True, index_col = 'Date')
df_returns.drop(columns = ['Unnamed: 0'], inplace = True)
df_returns.index = df_returns.index.strftime('%Y-%m-%d')
MSFT = pd.DataFrame(df_returns.loc[df_returns['Instrument'] == 'MSFT.O', 'Price Close'])
```

# Numpy and Pandas Functions

## Numpy

### np.where(), np.any()

np.where(condition, x, y) : returns a True/False array for an array if the condition is True, or return x (if true) or y (if false)

Use with df.apply() -> function to be applied must take in series/array as argument. Specify limits as well 

```python
# np.where()
MSFT2 = MSFT.copy()
MSFT2['Where Row'] = np.where(MSFT2['Price Close'] > 101, MSFT2['Price Close'] - 200, MSFT2['Price Close'] + 200)
print(MSFT2.head())

# Apply with Custom Function
def some_function(series, limit):
    # Use mask because we are dealing with series
    mask = series < limit
    # Use .where(mask)
    return series.where(mask, 'Greater than 100')

MSFT2.apply(some_function, limit = 100).head()

# Use Grid
grid_conditions = [[ lambda x : x == 'A', False],
                   [False, True]]
if_true = [['A', 'A'],
           ['B', 'B']]

if_false = [['1', '1'],
            ['2', '2']]

np.where(grid_conditions, if_true, if_false)
```

##### np.any() and np.all()

Returns True if any element is actually true, otherwise returns False
Specify the axis
Can skip NaN values can use skipna = True
Can do for entire dataframe (aggegrate over entire df) using axis = None

##### np.all()
Same idea but just for all

```python
Zero_df = pd.DataFrame(np.zeros(9).reshape(3,3))
Zero_df['1 Row'] = np.ones(3)
print(Zero_df)

# Across columns (reduces index)
print('\n Any Axis = 0: ', Zero_df.any(axis = 0))
# Across rows (reduces columns)
print('\n All Axis = 1:', Zero_df.all(axis = 1))

# Aggegrate (reduces columns and index to scale)
print('\n Any Axis = None: ', Zero_df.any(axis = None))
print('\n All Axis = None: ', Zero_df.all(axis = None))
```

## Pandas

### Iterrows and items

```python
# Zips a row-column(s) into a pair. Is a generator, use next() to get the next value
print('Index-Series value pairs : \n', next(MSFT2.iterrows()))

# Zips a row and iterates over column name, Series pairs
print('\nColumn-Series Pairs: : \n', next(MSFT2.items()))
```
