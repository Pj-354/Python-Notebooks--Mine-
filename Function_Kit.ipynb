$$ Geometric \; Brownian \; Motion $$
We can compact the random generations by centering the distribution at $\mu*dt$ and scaling it by $\sigma \sqrt{dt}$ directly, and also directly cumprod() the dataframe

```python

def geometric_brownian_return_efficient(n_years = 10, scenarios = 1000,
                              drift=0.07, volatility=0.15, 
                              dt_steps_per_year = 12,
                              s_0 = 100, prices = True):
    
    #Steps : number of years * steps per years

    dt = 1/dt_steps_per_year
    number_of_steps = int(n_years * dt_steps_per_year)

#We want to generate matrix of random numbers for the Brownian component
#Number of row is the new stock price, and every column is each scenario

    returns = np.random.normal(loc= (1+ drift )** dt, scale = (volatility * np.sqrt(dt)), 
                          size=(number_of_steps, scenarios))

#Make sure all returns start at s_0 by setting first return to 1 
    returns[0] = 1
    
#Convert returns to prices
    prices = s_0 * pd.DataFrame(returns).cumprod() if prices else returns-1
    return prices
```

```python
def plot_GBM(scenarios, drift, volatility):
    'Uses GBM Function from this file'
    s_0 = 100
    prices = geometric_brownian_return_efficient(
        scenarios=scenarios, drift=drift, volatility=volatility, s_0=s_0)
    #Plotting the Price Evolution
    ax = prices.plot(legend = False, alpha = 0.5, linewidth =2, figsize=(12,6))
    ax.axhline(y = 100, ls=':', color='black')
    #Set the Y-axis so it doesn't change 
    ax.set_ylim(top = 1000, bottom = -100)
    #Draw the plot with dot at origin
    ax.plot(0,s_0, marker='o',color='darkred', alpha=0.2)

## This creates the panel that you can change parameters for. Pass the function 
## to make interactive as the first argument, and next arguements use a tuple to 
## specify start, end, step
```

```python
def run_cppi(risky_r, safe_r=None, m=3, start=1000, floor=0.8, riskfree_rate=0.03, drawdown=None):
    """
    Run a backtest of the CPPI strategy, given a set of returns for the risky asset
    Returns a dictionary containing: Asset Value History, Risk Budget History, Risky Weight History
    """
    # set up the CPPI parameters
    dates = risky_r.index
    n_steps = len(dates)
    account_value = start
    floor_value = start*floor
    peak = account_value
    if isinstance(risky_r, pd.Series): 
        risky_r = pd.DataFrame(risky_r, columns=["R"])

    if safe_r is None:
        safe_r = pd.DataFrame().reindex_like(risky_r)
        safe_r.values[:] = riskfree_rate/12 # fast way to set all values to a number
    # set up some DataFrames for saving intermediate values
    account_history = pd.DataFrame().reindex_like(risky_r)
    risky_w_history = pd.DataFrame().reindex_like(risky_r)
    cushion_history = pd.DataFrame().reindex_like(risky_r)
    floorval_history = pd.DataFrame().reindex_like(risky_r)
    peak_history = pd.DataFrame().reindex_like(risky_r)

    for step in range(n_steps):
        if drawdown is not None:
            peak = np.maximum(peak, account_value)
            floor_value = peak*(1-drawdown)
        cushion = (account_value - floor_value)/account_value
        risky_w = m*cushion
        risky_w = np.minimum(risky_w, 1)
        risky_w = np.maximum(risky_w, 0)
        safe_w = 1-risky_w
        risky_alloc = account_value*risky_w
        safe_alloc = account_value*safe_w
        # recompute the new account value at the end of this step
        account_value = risky_alloc*(1+risky_r.iloc[step]) + safe_alloc*(1+safe_r.iloc[step])
        # save the histories for analysis and plotting
        cushion_history.iloc[step] = cushion
        risky_w_history.iloc[step] = risky_w
        account_history.iloc[step] = account_value
        floorval_history.iloc[step] = floor_value
        peak_history.iloc[step] = peak
    risky_wealth = start*(1+risky_r).cumprod()
    backtest_result = {
        "Wealth": account_history,
        "Risky Wealth": risky_wealth, 
        "Risk Budget": cushion_history,
        "Risky Allocation": risky_w_history,
        "m": m,
        "start": start,
        "floor": floor,
        "risky_r":risky_r,
        "safe_r": safe_r,
        "drawdown": drawdown,
        "peak": peak_history,
        "floor": floorval_history
    }
    return backtest_result
```

$$ Basic \; Summary \; Statistics $$

Functions include
1. Compound set of returns  
2. Annualising Returns
3. Annualising Volatility
4. Computing Semideviation
5. Skew & Kurtosis
6. Summary Statistics (returns dataframe of stats for a table of time series returns)

```python
def compound(r):
    """
    returns the result of compounding the set of returns in r
    """
    return np.expm1(np.log1p(r).sum())

def annualised_returns(r, periods):
    compounded_growth = (1 + r).prod()
    n = r.shape[0]
    return compounded_growth ** (periods / n) - 1

def annualised_vol(r, periods):
    vol = r.std()
    return vol * (periods**0.5)

def semideviation(strat = pd.Series):
    negative_returns = strat < 0 
    standard_deviation = strat.loc[negative_returns].std()
    return standard_deviation

def skew(strat = pd.Series):
    skewness = scipy.stats.skew(strat)
    return skewness

def kurt(strat = pd.Series):
    kurt = scipy.stats.kurtosis(strat)
    return kurt
```

```python
def drawdown(r):
    #Compute wealth over time (cumulative returns)
    wealth_level = 1000 * (1 + r).cumprod()
    peak = wealth_level.cummax()
    drawdown = (wealth_level - peak) / peak

    output = pd.DataFrame(  {'Wealth' : wealth_level,
                            'Peaks' : peak,
                            'Drawdown' : drawdown}
                            )
    return output



```
